Trabajo Práctico — Reconstrucción y regularización en imágenes tomográficas

## Ejercicio 2a

```python
def mascara_rayo(n, direccion, s){
    alpha,beta = direccion
    matriz= np.zeros((n,n), dtype=int)
    for i in range(n){
        for j in range (n){
            if alpha * i + beta*j == s{
                matriz[i][j]=1
            }
        }
    }
    return matriz
}
```
